The device operates using a $0^o : 45^o$ geometry, where the fading beam is projected vertically onto the sample, and the reflected radiation is collected from the side at an angle of 45 degrees (Figure 2, option 1). This is a common configuration in microfading devices, as it is relatively simple and requires only a single light source for both fading the sample and performing reflectance measurements. However, this design implies that the reflectance measurements are dependent on the fading light source. This can be problematic if the fading lamp is not suitable for accurate reflectance measurements. While this is not an issue when using a broadband white LED, it becomes a limitation when using a narrowband LED — an option available with the Fotonowy system — as it can seriously affect the quality of the color measurements. In our case, we purchased the scientific grade device which contains the possibility to choose among a 6 different LED lamps. We selected 2 broadband LEDs (warm and cold white) and 4 narrowband ones. So far, we have only used the broadband LEDs.

The only way to overcome this physical limitation is to modify the device’s design to include two separate light sources: one dedicated to the fading process and another to the color measurement process. The simplest solution would be to maintain the traditional $0^o : 45^o$ geometry but use a bifurcated fiber to alternate between two lamps, controlled by a TTL (transistor-transistor logic) function (Figure 2, option 2). Option 3, which is more complex and expensive, involves adding multiple dedicated measurement light sources as well as a camera. This is the approach we have implemented in our own stereo-MFT device (See Patin et al., 2022, *Journal of Cultural Heritage*, 57).

# Data
## Rawdata description 

Each microfading analyses performed with the Fotonowy device returns 4 files:
* a **.txt** file: it contains the CIE colorimetric values ($\Delta E$ and $L^*a^*b^*$) as a function of time and energy, as well as some metadata
* a **.rfc** file: it contains the inital and final reflectance spectra
* a **-spect.txt** file: it contains the raw signals (counts) recorded by the spectrometer (white, dark references and measurements on object) as a function of wavelengths
* a **-spect_convert.txt** file: same as the *-spect.txt* file but with a reduced wavelength range (380-780 nm)

The raw files can be opened with any basic text editors, but the readability of the content is not really high, especially for the spectral data files. This implies that for any visualizations or data manipulations, you are either uses the Fotonowy software which might be enough for some users or you implement your own solution which can range from an excel file to programming scripts.

## Raw files processing

This section describes our methodology to process the Fotonowy raw files into excel files. To perform this operation, we use the [microfading](https://www.pypi.org/) python package. We felt the need to convert the Fotonowy raw files into excel files for the following reasons:

- It enables us to create a file format which eases reading and access to the data.
- It enables us to add information about the object and project.
- It allows us to perform post-processing steps that are either not proposed by the Fotonowy software or not exactly the way we want. For example, it can be visualizations, interpolation, fitting, derivation, etc.
- Finally, using programming tools and developing a python package also allow us to automate certain tasks.

```python
# First we import the microfading package
import microfading as mf
```

```python
# Then we select the Fotonowy raw files
# In this case, we are the using the files provided by the microfading package, but ideally we should use your own files
rawFiles = mf.get_datasets(rawfiles=True, BWS=False)
rawFiles
```

```python
# Then, we process the raw data files
# By default the package save the excel file in the current working directory
# In this example, we set the "db" parameter to False
# But we recommend you to pre-register your object's info inside the database package so that you add the object metadata to the excel file.

interimFile = mf.process_rawdata(files=rawFiles, device='Fotonowy', db=False)
interimFile
```

## Data inspection 

This subsection focuses on the quality of the microfading data recorded for which we have several remarks:

1. Though the spectrometer seems to be able to record signal over a large wavelength range (90 - 800 nm), in practice only the signal between 420 and 700 nm can be used. While capturing signal over a larger wavelength range might be interesting in the context of research-oriented analyses, it is more than enough to compute color changes. Therefore it does not impact our ability to assess the light sensitivity of objects.  

2. In comparison with the stereo-MFT, the data obtained with the Fotonowy device are a bit more noisy. It does not prevent us to perform analyses, but it has several drawbacks:
    - You might need to use smoothing algorithms (Gaussian filter, Savitzky-Golay filter, etc.), at least for the sake of readibility when plotting several analyses together.
    - When the noise is too high, it becomes difficult to measure objects with a low light sensitivity.

3. There are multiple procedures to characterize the uncertainty of colour measurement devices (Early and Nadal, 2004). One of the simplest one is calculate the mean color difference to the mean (MCDM), which can be calculated for any colour difference equations ($\Delta E_{ab}$, $\Delta E_{94}$, $\Delta E_{00}$, etc.) (Billmeyer and Alessi, 1981). This can be done by performing successive color measurements on a known light stable material.


4. When performing microfading analyses on the fotolon sample provided by Fotonowy, we came accross some unexpected results in the spectral data. When looking at the spectral changes, we observed two peaks around 450 and 470 nm (Figure 3). This issue, also encountered by other users (see [Fotonowy forum](https://forum.fotonowy.pl/d/235-systematic-absorbance-band-appearance-450-and-470-nm/2)) for an answer has been provided by Fotonowy.

# Research use

In the framework of our laboratory, we are mainly using the Fotonowy device to provide microfading services to cultural institutions where the objective is to assess the light sensitivity of objects. It can be viewed as a conservation-oriented use of the device, which perfectly fullfils our needs. All the complex spectral and colorimetric aspects of microfadeometry are set aside as it is not needed to satisfy the light sensitivity requests of conservators and curators. On the other side, we do not use the device for research purposes. Adapting the device to meet such standard will require consequential investment and the market seems to be too small to be economically viable. Nevertheless, we listed a few comments that would help improving the research aspects of the device.

- Shift from a colorimetric-oriented system to a spectral-based system.
- Improve the quality of the recorded signal: reduce the noise and increase the wavelength range. 
- Add a proper imaging system, so that images can be used as computational data.
- Make the fading process and the collection process independent. This implies an extra light source.
