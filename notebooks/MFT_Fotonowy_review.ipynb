The device operates using a $0^o : 45^o$ geometry, where the fading beam is projected vertically onto the sample, and the reflected radiation is collected from the side at an angle of 45 degrees (Figure 2, option 1). This is a common configuration in microfading devices, as it is relatively simple and requires only a single light source for both fading the sample and performing reflectance measurements. However, this design implies that the reflectance measurements are dependent on the fading light source. This can be problematic if the fading lamp is not suitable for accurate reflectance measurements. While this is not an issue when using a broadband white LED, it becomes a limitation when using a narrowband LED — an option available with the Fotonowy system — as it can seriously affect the quality of the color measurements. In our case, we purchased the scientific-grade version of the device, which includes six interchangeable LEDs. We selected two broadband LEDs — warm white (2700 K) and cool white (5700 K) — along with four narrowband options. To date, we have primarily used the broadband LEDs.

The only way to overcome the limitation induced by the use of narrowband LEDs is to modify the device’s design to include two separate light sources: one dedicated to the fading process and another to the color measurement process. The simplest solution would be to maintain the traditional $0^o : 45^o$ geometry but use a bifurcated fiber to alternate between two lamps, controlled by a TTL (transistor-transistor logic) function (Figure 2, option 2). Option 3, which is more complex and expensive, but it gives you the possibility to add multiple dedicated measurement light sources as well as a high-quality camera. This is the approach we have implemented in our own stereo-MFT device (See Patin et al., 2022, *Journal of Cultural Heritage*, 57).

# Data
## Rawdata description 

Each microfading analyses performed with the Fotonowy device returns 4 files:
* a **.txt** file: it contains the CIE colorimetric values ($\Delta E$ and $L^*a^*b^*$) as a function of time and energy, as well as some metadata
* a **.rfc** file: it contains the inital and final reflectance spectra
* a **-spect.txt** file: it contains the raw signals (counts) recorded by the spectrometer (white, dark references and measurements on object) as a function of wavelengths
* a **-spect_convert.txt** file: same as the *-spect.txt* file but with a reduced wavelength range (380-780 nm)

The raw files can be opened in basic text editors, but the spectral data files, in particular, are not very readable. As a result, users must either rely on Fotonowy’s software—which may be sufficient for some—or implement their own solutions, which can range from spreadsheets to custom programming scripts.

## Raw files processing

This section describes our methodology to process the Fotonowy raw files into excel files. To perform this operation, we use the [microfading](https://www.pypi.org/) python package. We felt the need to convert the Fotonowy raw files into excel files for the following reasons:

- It enables us to create a file format which eases reading and access to the data.
- It enables us to add information about the object and project.
- It allows us to perform post-processing steps that are either not proposed by the Fotonowy software or not exactly the way we want. For example, it can be visualizations, interpolation, fitting, derivation, etc.
- Finally, using programming tools and developing a python package also allow us to automate certain tasks.

```python
# First we import the microfading package
import microfading as mf
```

```python
# Then we select the Fotonowy raw files
# In this case, we are the using the files provided by the microfading package, but ideally we should use your own files
rawFiles = mf.get_datasets(rawfiles=True, BWS=False)
rawFiles
```

```python
# Then, we process the raw data files
# By default the package save the excel file in the current working directory
# In this example, we set the "db" parameter to False
# But we recommend you to pre-register your object's info inside the database package so that you add the object metadata to the excel file.

interimFile = mf.process_rawdata(files=rawFiles, device='Fotonowy', db=False)
interimFile
```

## Data inspection 

This subsection focuses on the quality of the microfading data collected, for which we have several observations:

1. lthough the spectrometer is technically capable of recording across a broad wavelength range (approximately 90–800 nm), in practice, only the 420–700 nm region is usable. While a broader range may be desirable for research-oriented work, this limitation does not hinder our ability to calculate color changes or assess light sensitivity.

2. Compared to the stereo-MFT, the data obtained with the Fotonowy device exhibits slightly more noise. While this does not prevent analysis, it introduces several drawbacks:
    - Smoothing algorithms (e.g., Gaussian filter, Savitzky-Golay filter) may be needed — at least to improve readability when visualizing multiple measurements.
    - Excessive noise can complicate the measurement of objects with low light sensitivity.

# Research use

In our laboratory, the device is primarily used to offer microfading services to cultural institutions, where the goal is to assess the light sensitivity of objects. This is a conservation-oriented use case, and the device meets our needs well. In this context, complex spectral and colorimetric aspects of microfadeometry are largely irrelevant, as they are not necessary for satisfying the requirements of conservators and curators.

Conversely, we do not use the device for research purposes. Adapting it for such use would require significant investment, and the market for this application may be too limited to justify the cost. Nevertheless, we have identified a few suggestions that could improve the device’s research capabilities:

- Shift from a colorimetric-oriented system to a spectral-based system.
- Improve the signal quality by reducing noise and expanding the usable wavelength range. 
- Integrate a proper imaging system so that images can be used as computational data.
- Decouple the fading and measurement light sources to allow independent control. This implies an extra light source.
